**SETUP**

Follow the below instructions before running this program...

1) Create a google drive folder called YOLO.

2) Within the YOLO folder create a subfolder with a unique name and add your training data in it. Training data should be a .zip file names as images.zip

3) Make sure images.zip contains the following: .png images and their corresposding .txt files. Also, a classes.txt file containing the labels used for the training data should be present.


4) Add your rclone.conf file to the current directory - make sure rclone is setup to use google drive and name the the section as **[my_drive]
**

5) At the end you will have to add the unique sub folder name (which contains images.zip) in the 'drive_folder' variable.
